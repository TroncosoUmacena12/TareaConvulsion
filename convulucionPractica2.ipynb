```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convulsion_media_horizontal(imagen, tamañoKernel):
    kernel = np.ones([1, tamañoKernel], np.float32)/tamañoKernel
    img_fin = cv2.filter2D(imagen, -1, kernel)
    return img_fin 

def convulsion_media_vertical(imagen, tamañoKernel):
    kernel = np.ones([tamañoKernel,1], np.float32)/tamañoKernel
    img_fin = cv2.filter2D(imagen, -1, kernel)
    return img_fin 

def convulsion_media_diagonal(imagen, tamañoKernel):
    kernel = np.zeros((tamañoKernel,tamañoKernel), np.float32)
    np.fill_diagonal(kernel, 1.0 / tamañoKernel)
    img_fin = cv2.filter2D(imagen, -1, kernel)
    return img_fin



imagen = cv2.imread('P1.png')
gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)

img_rest = convulsion_media_horizontal(gris,10)
img_rest2 = convulsion_media_vertical(gris,10)
img_rest23 = convulsion_media_diagonal(gris,25) 

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.title('Imagen Original')
plt.imshow(cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB))

plt.subplot(1, 2, 2)
plt.title('Suavizado Horizontal')
plt.imshow(img_rest, cmap='gray')

plt.figure(figsize=(10,5))

plt.subplot(1, 2, 1)
plt.title('Suavizado Vertical')
plt.imshow(img_rest2, cmap='gray')

plt.subplot(1, 2, 2)
plt.title('Suavizado Diagonal')
plt.imshow(img_rest23, cmap='gray')


plt.show()
```
