Contraste

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def contraste(imagen, factor=1.0):
    img = imagen.astype(np.float32)
    media = np.mean(img)
    img_contraste = media + (img - media) * factor
    img_contraste = np.clip(img_contraste, 0, 255).astype(np.uint8)
    return img_contraste

img = cv2.imread('2.png', 0)

factor_contraste = 4
img_nueva = contraste(img, factor_contraste)

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].imshow(img, cmap='gray')
ax[0].set_title('Imagen Original')
ax[0].axis('off')

ax[1].imshow(img_nueva, cmap='gray')
ax[1].set_title('Imagen con Contraste Ajustado')
ax[1].axis('off')

plt.show()
```

contraste a colores

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def contraste_color(img, factor=1.0):
    img = img.astype(np.float32)
    for i in range(3):
        media = np.mean(img[:,:,i])
        img[:,:,i] = media + (img[:,:,i] - media) * factor
        img[:,:,i] = np.clip(img[:,:,i],0,255)
    return img.astype(np.uint8)

img = cv2.imread('2.png')
img_constraste = 15
img_nueva = contraste_color(img_nueva,img_constraste)
plt.figure(figsize=(10,5))

plt.subplot(121)
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title('Imagen original')

plt.subplot(122)
plt.imshow(cv2.cvtColor(img_nueva,cv2.COLOR_BGR2RGB))
plt.title('Imagen negativo')

plt.show()
```

Histograma de una imagen con contrastado

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def contraste(imagen, factor=1.0):
    imagen = imagen.astype(np.float32)
    media = np.mean(imagen)
    contrastado = media + (imagen - media) * factor
    contrastado = np.clip(contrastado, 0, 255).astype(np.uint8)
    return contrastado

img = cv2.imread('2.png', 0)

f_contraste = 5
img_nueva = contraste(img, f_contraste)

histograma_or = cv2.calcHist([img], [0], None, [256], [0, 256])
histograma_nuevo = cv2.calcHist([img_nueva], [0], None, [256], [0, 256])

plt.figure(figsize=(14, 5))

plt.subplot(221)
plt.imshow(img, cmap='gray')
plt.title('Imagen original')
plt.axis('off')

plt.subplot(222)
plt.plot(histograma_or, color='black')
plt.title('Histograma original')
plt.xlabel('Intensidad')
plt.ylabel('Frecuencia')

plt.subplot(223)
plt.imshow(img_nueva, cmap='gray')
plt.title('Imagen original')
plt.axis('off')

plt.subplot(224)
plt.plot(histograma_nuevo, color='red')
plt.title('Histograma imagen con contraste')
plt.xlabel('Intensidad')
plt.ylabel('Frecuencia')

plt.tight_layout()
plt.show()
```

Ecualizacion del un Histrograma - Se desea normalizar 

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
 
imagen = cv2.imread('2.png',0)
 
histograma,calc=np.histogram(imagen.flatten(),256,[0,256])
distriAcu = histograma.cumsum()
distriAcu_normalizada = distriAcu*histograma.max()/distriAcu.max()
ecualizado = cv2.equalizeHist(imagen)
histogrma_equ,_=np.histogram(ecualizado.flatten(),256,[0,256])
 
plt.figure(figsize=(10,5))
plt.subplot(2,2,1)
plt.imshow(imagen,cmap='gray')
plt.title('imagen original')
plt.axis('off')
 
plt.subplot(2,2,2)
plt.plot(distriAcu_normalizada,color='b')
plt.hist(imagen.flatten(),256,[0,256],color='r')
plt.xlim([0,256])
plt.legend(('distribucion','histograma'),loc='upper left')
plt.title('histograma original')
 
 
plt.subplot(2,2,3)
plt.imshow(ecualizado,cmap='gray')
plt.title('imagen ecualizada')
plt.axis('off')
 
plt.subplot(2,2,4)
plt.plot(histogrma_equ,color='b')
plt.xlim([0,256])
plt.tight_layout()
 
plt.show()
```
